```python
def ingest_db(file_name, table_name, db_engine):
    '''this fun will ingest the dataframe into database table'''
    for chunk in pd.read_csv(file_name, chunksize=100000):
        chunk.to_sql(table_name, db_engine,
                     if_exists='append',
                     index=False)
    print(f"Ingested {table_name} into database")


import os 
import pandas as pd
import time  # Added missing import for time
import logging  # Added missing import for logging

df = None 
# Need to define engine or pass it to the function
# engine = your_database_engine_here

from sqlalchemy import create_engine

engine = create_engine('sqlite:///inventory.db')

def load_raw_data():
    '''this fun will load the csvs as dataframe and ingest into db'''
    start = time.time()  # Moved inside function
    
    for file in os.listdir():  # Indented to be inside the function
        if file.endswith('.csv'):
            logging.info(f'Ingesting {file} in db')
            ingest_db(file, file[:-4], engine)
    
    end = time.time()  # Moved inside function
    
    total_time = (end - start)/60  # Moved inside function
    logging.info('Ingestion Complete')
    logging.info(f'\nTotal Time Taken: {total_time} minutes')

if __name__ == '__main__':
    load_raw_data()
```
